```python
import numpy as np

'''
An HR system stores employee performance ratings (scale 1–5) 
for 100 employees over 4 quarters in a NumPy array.

'''
ratings=np.random.randint(1,6,size=(100, 4))

#Min_max normalization

min_val=np.min(ratings)
max_val=np.max(ratings)
normalized_ratings = (ratings - min_val) / (max_val - min_val)

# Average rating per employee
employee_avg = np.mean(normalized_ratings, axis=1)
print(f"Employee Average {employee_avg}")
company_mean = np.mean(employee_avg)
print(f"Company mean {company_mean}")
top_employees = np.where(employee_avg > company_mean)
print(f"Top Employees {top_employees}")
```
