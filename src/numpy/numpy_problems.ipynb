5. create a 2D array with 1 on the border and 0 inside

```python
import numpy as np
```

```python
import numpy as np

arr = np.zeros(25).reshape(5, 5)
arr[0, :] = 1  #first row
arr[-1, :] = 1  #last row
arr[:, 0] = 1  #first col
arr[:, -1] = 1  #last col
print(arr)
```

6. Write a Python program to add a border (filled with 0's) around an existing array

```python
import numpy as np

arr = np.ones(25).reshape(5, 5)
arr[0, :] = 0
arr[-1, :] = 0
arr[:, 0] = 0
arr[:, -1] = 0
print(arr)
```

7. Write a Python program to create a 8x8 matrix and fill it with a checkerboard pattern

```python
import numpy as np

arr = np.zeros(64).reshape(8, 8)
arr[::2, 1::2] = 1
arr[1::2, ::2] = 1
print(arr)
```

10. Write a Python program to find the real and imaginary parts of an array of complex
numbers.

```python
import numpy as np

arr = np.array([1.00000000 + 0.j, 0.70710678 + 0.70710678j])
print(f"Real part of array:{np.real(arr)}")
print(f"Imaginary part of array:{np.imag(arr)}")
```

11. Write a Python program to find the number of elements of an array, length of one
array element in bytes and total bytes consumed by the elements.
Expected Output:
Size of the array: 3
Length of one array element in bytes: 8
Total bytes consumed by the elements of the array: 24

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])
print(f"Size of the array:{len(arr)}")
print(f"Length of the array element in bytes: {arr.itemsize}")
print(f"Total bytes consumed by elements of the array: {arr.nbytes}")
```

12. Write a Python program to find common values between two arrays.
Expected Output:
Array1: [ 0 10 20 40 60]
Array2: [10, 30, 40]
Common values between two arrays:
[10 40]

```python
import numpy as np

arr1 = np.array([0, 10, 20, 40, 60])
arr2 = np.array([10, 30, 40])
print(np.intersect1d(arr1, arr2))
```
