```python
import pandas as pd
from faker import Faker
fake=Faker()
data=[]
for i in range(10):
    data.append({
        'order_id':i+1,
        'customer_id':fake.random_int(min=1000,max=1020),
        'order_date':fake.date_between(start_date='-1y', end_date='today'),
        'order_amount':round(fake.pyfloat(min_value=100,max_value=5000),2)
    })

df=pd.DataFrame(data)

print(df)
```

```python
print(df.info())
```

```python
df["order_date"]=pd.to_datetime(df["order_date"])
print(df.info())
```

```python
monthly_revenue=df.groupby(df["order_date"].dt.to_period('M'))['order_amount'].sum()
print(monthly_revenue)
```

```python
unique_customers=df.groupby(df["order_date"].dt.to_period("M"))["customer_id"].nunique()
print(unique_customers)

print(unique_customers.idxmax())
```
