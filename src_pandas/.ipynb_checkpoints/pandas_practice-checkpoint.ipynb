Pandas Basics

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40, 50, 60])
print(s)
print(s.dtype)
print(s.values)
print(s.index)
s.name = "numbers"
print(s)
```

creating series from dictionary

```python
import pandas as pd
fruit_protein={
    'Avocado':2.0,
    'Guava':2.6,
    'BlackBerries':2.0,
    'Oranges':0.9,
    'Banana':1.1,
    'Mango':0.8
}

s2=pd.Series(fruit_protein,name="Protein")
print(s2)

#conditional selection
print(s2 > 1.0)
print(s2[s2 > 1.0])

```

Pandas DataFrames

```python
import numpy as np
import pandas as pd
data={
    "Name":["Alice","Bob","Charlie","Sowmya","Suji","Bhavana"],
    "Age":[25,30,35,np.nan,21,22],
    "Department":["HR","IT","FINANCE","IT","HR","HR"],
    "SALARY":[50000,60000,70000,62000,np.nan,50000]
}
print(data)
df=pd.DataFrame(data)
print(df)

df.tail()

print(df.iloc[1:3, :2])

df.drop("Age", axis=1)
```
