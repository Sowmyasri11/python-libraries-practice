```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
x=[1,2,3,4]
y=[5,6,7,8]
plt.plot(x,y)
plt.grid()
plt.show()
```

```python
#pyplot API

#Univariate -Numerical
data={
    "salary":[25000,30000,37000,28000,39000,48000,55000,52000,35000,31000]
}
df=pd.DataFrame(data)
print(df)
```

```python
#Line Plot:

plt.plot(df["salary"])
plt.show()
```

```python
# Matplotlib Markers
```

```python
# Mark each point with a circle:

plt.plot(df["salary"],marker='o')
```

```python
# Format Strings fmt
'''
written with this syntax:

marker|line|color
'''
plt.plot(df["salary"], 'o:r')
plt.show()
```

```python
# marker size - ms

plt.plot(df["salary"],marker='o',ms=20)
```

```python
# markeredgecolor - mec

plt.plot(df["salary"],marker='o',ms=20, mec='r')
```

```python
# markerfacecolor - mfc

plt.plot(df["salary"],marker='o',ms=20, mec='b', mfc='hotpink')
```

```python
plt.plot(df["salary"],linestyle='dashed')
```

```python
'''
linestyle can be written as ls.

dotted can be written as :.

dashed can be written as --.
'''
```

```python
# Linecolor -c
plt.plot(df["salary"],color='green')
```
